# 1. Data Exploration & Analysis (EDA)

This notebook explores the Alzheimer's disease biomarker dataset to understand its structure, distributions, and relationships.

## Setup and Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

print('Libraries loaded successfully')
```

## Load Raw Data

```python
# Load raw data
df = pd.read_csv('../data/raw/alzheimers_biomarkers.csv')

print(f'Dataset shape: {df.shape}')
print(f'\nFirst 5 rows:')
print(df.head())
```

## Data Types and Info

```python
# Data types and info
print('Data Types:')
print(df.dtypes)
print('\nDataset Info:')
print(df.info())
```

## Descriptive Statistics

```python
# Descriptive statistics
print('Descriptive Statistics:')
print(df.describe())
```

## Missing Values Analysis

```python
# Check for missing values
missing = df.isnull().sum()
print('Missing Values:')
print(missing)
print(f'\nTotal missing: {missing.sum()}')
```

## Target Variable Distribution

```python
# Target variable distribution
target_dist = df['amyloid_positive'].value_counts()
print('Target Distribution:')
print(target_dist)
print(f'\nPercentage:')
print(df['amyloid_positive'].value_counts(normalize=True) * 100)

# Plot
plt.figure(figsize=(8, 5))
df['amyloid_positive'].value_counts().plot(kind='bar', color=['#1f77b4', '#ff7f0e'])
plt.xlabel('Amyloid Status')
plt.ylabel('Count')
plt.title('Target Variable Distribution')
plt.xticks([0, 1], ['Negative', 'Positive'], rotation=0)
plt.tight_layout()
plt.show()
```

## Feature Distributions

```python
# Plot distributions of all features
features = ['p_tau', 'np_tau', 'ab_42', 'age', 'years_education', 'mmse']

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.ravel()

for idx, feature in enumerate(features):
    axes[idx].hist(df[feature], bins=30, color='steelblue', edgecolor='black', alpha=0.7)
    axes[idx].set_xlabel(feature)
    axes[idx].set_ylabel('Frequency')
    axes[idx].set_title(f'Distribution of {feature}')
    axes[idx].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

## Box Plots by Target Variable

```python
# Box plots by target variable
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.ravel()

for idx, feature in enumerate(features):
    data_neg = df[df['amyloid_positive'] == 0][feature]
    data_pos = df[df['amyloid_positive'] == 1][feature]
    
    axes[idx].boxplot([data_neg, data_pos], labels=['Negative', 'Positive'])
    axes[idx].set_ylabel(feature)
    axes[idx].set_title(f'{feature} by Amyloid Status')
    axes[idx].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

## Correlation Analysis

```python
# Correlation matrix
corr_matrix = df.corr()

# Plot correlation heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()

# Correlations with target
print('\nCorrelations with Target (amyloid_positive):')
target_corr = corr_matrix['amyloid_positive'].sort_values(ascending=False)
print(target_corr)
```

## Feature Relationships with Target

```python
# Statistical tests (t-tests)
print('T-tests (Negative vs Positive):')
print('-' * 50)

for feature in features:
    neg = df[df['amyloid_positive'] == 0][feature]
    pos = df[df['amyloid_positive'] == 1][feature]
    
    t_stat, p_value = stats.ttest_ind(neg, pos)
    print(f'{feature:20} | t-stat: {t_stat:8.3f} | p-value: {p_value:.2e}')
```

## Summary Statistics by Target Group

```python
# Group by target
print('Summary Statistics by Amyloid Status:')
print('\nAmyloid Negative:')
print(df[df['amyloid_positive'] == 0][features].describe())

print('\nAmyloid Positive:')
print(df[df['amyloid_positive'] == 1][features].describe())
```

## Key Findings

1. **Dataset Size**: 500 samples with 60% positive, 40% negative class distribution
2. **No Missing Values**: Dataset is complete
3. **Feature Ranges**: All features within expected biological ranges
4. **Class Differences**: Significant differences between positive and negative groups across most features
5. **Correlations**: Strong correlations between biomarker features (p_tau, np_tau, ab_42)
6. **MMSE Separation**: Clear difference in cognitive scores between groups (lower for positive)
